```python

import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
df = pd.read_csv("../all_hrv_features.csv")   # Replace with your file name
print("Initial Shape:", df.shape)
```

```python
df.isnull().sum()[df.isnull().sum() > 0]
```

```python
df.duplicated().sum()
```

```python
X = df.drop(columns=['label'])
y = df['label']

# 4️⃣ Encode Target Labels
# Example: Normal, Abnormal, MI, History_MI, Arrhythmia, CAD
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)
print("\nLabel Mapping:", dict(zip(label_encoder.classes_, label_encoder.transform(label_encoder.classes_))))

# 5️⃣ Identify Feature Types
numeric_features = X.select_dtypes(include=[np.number]).columns.tolist()
categorical_features = X.select_dtypes(exclude=[np.number]).columns.tolist()

print("\nNumeric Features:", len(numeric_features))
print("Categorical Features:", len(categorical_features))
```

```python
from sklearn.preprocessing import FunctionTransformer
# 6️⃣ Numeric Preprocessing
numeric_pipeline = Pipeline(steps=[
    ('replace_inf', FunctionTransformer(lambda x: np.where(np.isinf(x), np.nan, x))), 
    ('imputer', SimpleImputer(strategy='constant', fill_value = 0)),       # Handle missing numeric values
    ('scaler', StandardScaler())                         # Standardize numeric features
])

# 7️⃣ Categorical Preprocessing (if any)
categorical_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
])

# 8️⃣ Combine Pipelines
preprocessor = ColumnTransformer(transformers=[
    ('num', numeric_pipeline, numeric_features),
    ('cat', categorical_pipeline, categorical_features)
])

```

```python
X[np.isinf(X.select_dtypes(include=[np.number])).any(axis=1)]
```

```python
# 9️⃣ Fit & Transform
X_clean = preprocessor.fit_transform(X)

# 🔟 Convert to Cleaned DataFrame
X_clean_df = pd.DataFrame(X_clean, columns=numeric_features + categorical_features)
X_clean_df['label'] = y_encoded

print("\nCleaned Data Shape:", X_clean_df.shape)
print("\nPreview:\n", X_clean_df.head())

# Optional: Save Cleaned Dataset
X_clean_df.to_csv("heart_hrv_cleaned.csv", index=False)
print("\n✅ Cleaned dataset saved as 'heart_hrv_cleaned.csv'")
```
